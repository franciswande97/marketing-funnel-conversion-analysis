```python
import pandas as pd

users = pd.read_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\raw\\users.csv")
ads = pd.read_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\raw\\ads.csv")
campaigns = pd.read_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\raw\\campaigns.csv")
ad_events = pd.read_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\raw\\ad_events.csv")
```

```python
#users csv EDA
users.info()
print(f"Number of null values in users DataFrame: \n{users.isnull().sum()}")
print(f"Number of duplicate rows in users DataFrame: {users.duplicated().sum()}")
users.head(5)
```

```python
#ads csv EDA
ads.info()
print(f"Number of null values in ads DataFrame: \n{ads.isnull().sum()}")
print(f"Number of duplicate rows in ads DataFrame: {ads.duplicated().sum()}")
ads.head(5)
```

```python
#campaigns csv EDA
campaigns.info()
print(f"Number of null values in campaigns DataFrame: \n{campaigns.isnull().sum()}")
print(f"Number of duplicate rows in campaigns DataFrame: {campaigns.duplicated().sum()}")
campaigns.head(5)
```

```python
#ads events csv EDA
ad_events.info()
print(f"Number of null values in ad_events DataFrame: \n{ad_events.isnull().sum()}")
print(f"Number of duplicate rows in ad_events DataFrame: {ad_events.duplicated().sum()}")
ad_events.head(5)
```

```python
campaigns["start_date"] = pd.to_datetime(campaigns["start_date"])

campaigns["end_date"] = pd.to_datetime(campaigns["end_date"])

campaigns.info()
```

```python
ad_events["timestamp"] = pd.to_datetime(ad_events["timestamp"])

ad_events["date"] = ad_events["timestamp"].dt.date

ad_events["month"] = ad_events["timestamp"].dt.month_name()

ad_events["hour"] = ad_events["timestamp"].dt.hour

ad_events.info()
ad_events.head(5)
```

```python
#saving changes to new csv files
users.to_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\processed\\users.csv", index=False)
ads.to_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\processed\\ads.csv", index=False)
campaigns.to_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\processed\\campaigns.csv", index=False)
ad_events.to_csv("C:\\Projects & Datasets\\Python\\Analysis App\\data\\processed\\ad_events.csv", index=False)
```
